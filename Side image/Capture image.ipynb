```python
import tkinter as tk
import cv2
import PIL.Image
import PIL.ImageTk

def capture_photo():
    global photo_image
    ret, frame = cap.read()
    if ret:
        cv2.imwrite("captured_image.jpg", frame)
        # Convert BGR to RGB for PIL
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        photo_image = PIL.Image.fromarray(frame)
        photo_image = photo_image.resize((400, 300))  # Adjust image size as needed
        photo_image = PIL.ImageTk.PhotoImage(photo_image)
        label_image.config(image=photo_image)
        label_message.config(text="Photo captured and saved!")
    else:
        label_message.config(text="Error capturing photo.")

root = tk.Tk()
root.title("Photo Capture")

label_preview = tk.Label(root)
label_preview.pack()

button_capture = tk.Button(root, text="Capture Photo", command=capture_photo)
button_capture.pack()

label_message = tk.Label(root, text="Click the button to capture a photo.")
label_message.pack(pady=10)

label_image = tk.Label(root)
label_image.pack()

cap = cv2.VideoCapture(0)

def update_preview():
    global photo_image
    ret, frame = cap.read()
    if ret:
        # Convert BGR to RGB for PIL
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        photo_image = PIL.Image.fromarray(frame)
        photo_image = photo_image.resize((400, 300))  # Adjust image size as needed
        photo_image = PIL.ImageTk.PhotoImage(photo_image)
        label_preview.config(image=photo_image)
    root.after(20, update_preview)  # Update every 20 milliseconds

update_preview()

root.mainloop()
cap.release()
```
