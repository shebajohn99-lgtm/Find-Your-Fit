```python
import cv2
import numpy as np
import os
import imutils
from matplotlib import pyplot as plt #import matplotlib
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split


# Function to extract features from an image
def extract_human_contour(image_path):
    img = cv2.imread(image_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    #_, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray = cv2.GaussianBlur(gray, (5, 5), 0)
    edged = cv2.Canny(gray, 35, 125)

    # Find contours
   # contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    contours, _ = cv2.findContours(edged.copy(), cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)

    # Find the largest contour    (assuming it's the human)
    largest_contour = max(contours, key=cv2.contourArea)
    x, y, w, h = cv2.boundingRect(largest_contour)

    # Crop the human region
    human_region = img[y:y+h, x:x+w]
    plt.imshow(human_region)
    plt.show()

    # Extract features from the human region
    height, width, _ = human_region.shape
    return [height, width]


# Function to load and preprocess images
def load_and_preprocess_images(image_folder):
    images = []
    labels = []
    for filename in os.listdir(image_folder):
        img_path = os.path.join(image_folder, filename)
        if is_hidden(img_path):
            print("hidden file")
        elif os.path.isfile(img_path) and not is_hidden(img_path):
            features = extract_human_contour(img_path)
            label = classify_image_size(img_path)  # Use a simple classification rule or a pre-trained model
        #    print("label", label)
            images.append(features)
            labels.append(label)

    return np.array(images), np.array(labels)
# Classify images
def classify_image_size(image_path):
    img = cv2.imread(image_path)
    height, width, _ = img.shape

    # Define size thresholds (adjust as needed)
    small_threshold = 800
    medium_threshold = 1100
  #  print ("width", width)
    if width < small_threshold:
        return "S"
    elif width < medium_threshold:
        return "M"
    else:
        return "L"
    
# Train the model
def train_model(X_train, y_train):
    model = SVC()  # You can experiment with other models like Random Forest, Neural Networks
    model.fit(X_train, y_train)
    return model

# Predict image size
def predict_size(model, image_path):
    features = extract_human_contour(image_path)
    print ("Predicted features", features)
    predicted_size = model.predict([features])[0]
    return predicted_size

# Example usage
def is_hidden(file):
    return file.startswith('.')

image_folder = 'Front image'
X, y = load_and_preprocess_images(image_folder)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = train_model(X_train, y_train)

new_image_path = "Front image/woman7.jpg"
predicted_size = predict_size(model, new_image_path)
print("Predicted size:", predicted_size)
```
